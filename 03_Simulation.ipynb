```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO
plt.rcParams.update({'figure.figsize': (9, 4), 'axes.grid': True})
```

## 1. Two models of growth
An exponential model assumes unlimited resources: **N(t) = N₀ exp(rt)**.

A logistic model includes a carrying capacity K: **N(t) = K / [1 + ((K − N₀)/N₀) exp(−rt)]**.

Here N is a modelled population, t is days and r is a growth-rate parameter in day⁻¹. The smooth curve is a simplified deterministic model, not a prediction of any real organism. We assume positive growth and an initial population no greater than K.

```python
def growth_models(initial_population, growth_rate, capacity, duration_days):
    values = [initial_population, growth_rate, capacity, duration_days]
    if not np.isfinite(values).all():
        raise ValueError('Use finite numeric parameters.')
    if initial_population <= 0 or growth_rate < 0 or capacity < initial_population or duration_days <= 0:
        raise ValueError('Require 0 < initial population <= capacity, rate >= 0 and duration > 0.')
    t = np.linspace(0, duration_days, 201)
    if growth_rate * duration_days > 600:
        raise ValueError('Rate × duration is too large for this demonstration. Reduce one of them.')
    exponential = initial_population * np.exp(growth_rate * t)
    logistic = capacity / (1 + ((capacity-initial_population)/initial_population) * np.exp(-growth_rate*t))
    return t, exponential, logistic
```

## 2. Run the simulation
Predict what doubling the rate will do **before** changing it. Then change one parameter at a time.

```python
# TODO: experiment with these four parameters.
initial_population = 20.0
growth_rate = 0.30       # per day
capacity = 500.0
duration_days = 15.0
t, exponential, logistic = growth_models(initial_population, growth_rate, capacity, duration_days)
fig, ax = plt.subplots()
ax.plot(t, exponential, label='Exponential')
ax.plot(t, logistic, label='Logistic')
ax.axhline(capacity, linestyle='--', color='grey', label='Carrying capacity')
ax.set(xlabel='Time (days)', ylabel='Modelled population', title='Two assumptions about population growth')
ax.legend()
fig.tight_layout()
fig.savefig('growth_models.png', dpi=160)
plt.show()
```

## 3. Check the model
Both curves should begin at N₀. Under these assumptions the logistic curve should never exceed K. With zero growth rate both populations should stay constant.

```python
assert np.isclose(exponential[0], initial_population)
assert np.isclose(logistic[0], initial_population)
assert (logistic <= capacity + 1e-8).all()
_, flat_exp, flat_log = growth_models(20, 0, 500, 15)
assert np.allclose(flat_exp, 20) and np.allclose(flat_log, 20)
results = pd.DataFrame({'time_days': t, 'exponential_population': exponential, 'logistic_population': logistic})
results.to_csv('growth_results.csv', index=False)
print('Checks passed. Final logistic population:', round(logistic[-1], 2))
```
